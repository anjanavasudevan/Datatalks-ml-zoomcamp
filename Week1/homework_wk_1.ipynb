## Matrix multiplication

```python
all_asian_cars = df.loc[df['origin']=='Asia'] # all asian cars only
display(all_asian_cars.head(10))

# select only vehicle weight and model year
fmatrix = all_asian_cars[['vehicle_weight', 'model_year']].values
print(type(fmatrix))
```

```python
# matrix multiplication on the required subset
import numpy as np
fsubset = fmatrix[:7, :]
print(fsubset.shape)

# transpose
fsubset_T = np.transpose(fsubset)
print(fsubset_T.shape)

# Matrix multiplication
xtx = fsubset_T @ fsubset
print(xtx.shape)

# Invert xtx
xtx_inv = np.linalg.inv(xtx)
print(xtx.shape)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = (xtx_inv @ fsubset_T) @ y
print(w.shape)
print(w.sum())
```
